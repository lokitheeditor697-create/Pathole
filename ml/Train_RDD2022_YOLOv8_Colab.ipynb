```python
# 3. Download RDD2022 (India Dataset)
# Option A: Via Kaggle API
# !kaggle datasets download -d sekilab/road-damage-dataset
# !unzip -q road-damage-dataset.zip -d ml/raw/RDD2022

# Option B: Via Direct Official RDD2022 Release Tarball
!mkdir -p ml/raw/RDD2022/India
!wget -q -O India.tar.gz "https://mycityreport.s3-ap-northeast-1.amazonaws.com/02_RoadDamageDataset/public_data/India.tar.gz" || echo "Use Kaggle or direct upload if AWS link requires token"
!tar -xzf India.tar.gz -C ml/raw/RDD2022/ || true
```

```python
# 4. Convert RDD2022 Pascal VOC XML to YOLOv8 format
!python ml/scripts/convert_rdd2022_to_yolo.py ml/raw/RDD2022 ml/datasets/rdd2022_yolo
```

```python
# 5. Train YOLOv8 with road-tuned augmentations
from ultralytics import YOLO

model = YOLO('yolov8s.pt')
results = model.train(
    data='ml/datasets/rdd2022_yolo/rdd2022.yaml',
    epochs=100,
    patience=25,
    imgsz=640,
    batch=16,
    optimizer='AdamW',
    lr0=0.001,
    lrf=0.01,
    warmup_epochs=3.0,
    hsv_h=0.015,
    hsv_s=0.7,
    hsv_v=0.4,
    fliplr=0.5,
    mosaic=1.0,
    close_mosaic=10,
    project='runs/train',
    name='rdd2022_yolov8'
)
```

```python
# 6. Evaluate model performance metrics
metrics = model.val()
print(f"mAP50: {metrics.box.map50:.4f}")
print(f"mAP50-95: {metrics.box.map:.4f}")
```
